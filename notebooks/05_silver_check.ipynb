```python
import pandas as pd

# ---------- CUSTOMERS ----------
bronze_cust = pd.read_csv("../data/lake/bronze/customers.csv")
silver_cust = pd.read_parquet("../data/lake/silver/customers.parquet")

print("Bronze customers (raw):", len(bronze_cust), "rows")
display(bronze_cust.head())

print("Silver customers (clean):", len(silver_cust), "rows")
# papar baris yang ada email sahaja supaya tiada None
display(silver_cust[silver_cust["email_missing"] == False].head())

# ---------- BOOKS ----------
books = pd.read_parquet("../data/lake/silver/books.parquet")
print("Silver books:", len(books), "rows")
display(books[["title", "price_gbp", "price_myr", "fx_rate_date", "rating"]].head())

# ---------- ORDERS ----------
orders = pd.read_parquet("../data/lake/silver/orders.parquet")
print("Silver orders:", len(orders), "rows")
display(orders.head())

# ---------- REVIEWS ----------
reviews = pd.read_parquet("../data/lake/silver/reviews.parquet")
print("Silver reviews:", len(reviews), "rows")
display(reviews.head())

# ---------- REVIEWS ----------
reviews = pd.read_csv("../data/lake/bronze/reviews.csv")
print("Bronze reviews:", len(reviews), "rows")
display(reviews.head())
```

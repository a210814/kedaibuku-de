Answer:

1. In books, one row is one book in the catalogue. In customers, one row is one customer. In orders, one row is one order, where a customer buys one book in a certain quantity on a certain date. In reviews, one row is one review written by a customer about a book.

2. The customer_id column links customers, orders and reviews, while the book_id column links books, orders and reviews.

3. The customers file holds personal data in full_name, email, phone and mykad, with MyKad being the most sensitive. The customer_id in orders and reviews is also personal data because it can be traced back to a person.

4. The book catalogue would be collected once, or rarely, because it changes very little. The BNM exchange rate would be collected every day because the rate changes daily. Orders and reviews would also be collected daily, or continuously in a real business, because new ones arrive all the time.

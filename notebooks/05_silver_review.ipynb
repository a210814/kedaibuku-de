Lab 5 (W5): Cleaning data and protecting personal data

1. Quarantine versus delete
Bad orders might be real sales with small mistakes, so deleting them loses revenue data. A separate file lets us check why they failed and fix them. The data engineer and data quality team would review it, and ask sales to confirm the valid orders.

2. Hash versus mask
Hashing gives the same output for the same email. Without a salt, someone can hash a list of common emails and match them to our email_hash. A secret salt stops this, but analysts can still count unique customers.

3. Data minimisation
This follows the PDPA 2010 General Principle: personal data should not be excessive for its purpose. Analytics does not need name, phone or MyKad. age_group is safer than birth_year because an exact year can help identify one person, but a group like 25-34 hides them among many.

4. Parquet versus CSV
When I checked `.dtypes` on `orders.parquet`, `quantity` was `int64` and `order_date` was `datetime64[us]`, while the ID columns were `object` (text). Parquet remembers these data types. CSV is just text, so `order_date` would be read as a plain string and I would have to convert it again every time I load the file.
